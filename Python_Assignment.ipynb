```python
# Replace 'not...poor' with 'good'
string = input("Enter a string: ")

not_pos = string.find("not")
poor_pos = string.find("poor")

if not_pos != -1 and poor_pos != -1 and not_pos < poor_pos:
    result = string[:not_pos] + "good" + string[poor_pos + 4:]
    print(result)
else:
    print(string)
```

```python
# Fibonacci series up to n terms
n = int(input("Enter n: "))

a, b = 0, 1

print("Fibonacci Series:")

for i in range(n + 1):
    print(a, end=" ")
    a, b = b, a + b
```

```python
# Sum of odd and even series
import math

def odd_series(n):
    total = 0
    for i in range(1, n + 1, 2):
        total += (i ** 2) / math.factorial(i)
    return total

def even_series(n):
    total = 0
    for i in range(2, n + 1, 2):
        total += (i ** 2) / math.factorial(i)
    return total

n = int(input("Enter n: "))

print("Odd series sum =", odd_series(n))
print("Even series sum =", even_series(n))
```

```python
# Factorial using recursion
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

num = int(input("Enter a number: "))

print("Factorial =", factorial(num))
```

```python
# Password Generator Mini Project
import random
import string

class User:
    def __init__(self, user_id, name, password):
        self.details = (user_id, name, password)

    def display(self):
        print("User Details:", self.details)

def generate_password(text):
    try:
        words = text.split()

        if len(words) == 0:
            raise ValueError("Please enter valid words.")

        selected = ''.join(random.sample(words, min(len(words), 2)))

        numbers = ''.join(random.choices(string.digits, k=2))
        special = ''.join(random.choices("!@#$%^&*", k=2))
        capital = random.choice(string.ascii_uppercase)

        password = selected + numbers + special + capital

        if len(password) < 8:
            password += ''.join(random.choices(string.ascii_letters, k=8-len(password)))

        return password

    except Exception as e:
        print("Error:", e)

user_id = input("Enter User ID: ")
name = input("Enter Name: ")
text = input("Enter words for password generation: ")

password = generate_password(text)

if password:
    user = User(user_id, name, password)
    user.display()
```
